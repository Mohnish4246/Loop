```python
# Print all prime numbers from 1 to 50
for num in range(2, 51):
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num,end=" ")
```

```python
# Count even & odd numbers in a list
nums = [10, 21, 4, 9, 16, 25]
even = 0
odd = 0

for n in nums:
    if n % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even:", even)
print("Odd:", odd)
```

```python
# Perfect squares between 1 and 200
for i in range(1, 201):
    if int(i**0.5) ** 2 == i:
        print(i,end=" ")
```

```python
s = input("Enter string: ")
upper = 0
lower = 0
for ch in s:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1
print("Uppercase:", upper)
print("Lowercase:", lower)
```

```python
s = input("Enter string: ") # reverse a string  (hello)
rev = ""
for ch in s:  # ch=hello
    rev = ch + rev   # rev = "h"+" = "h", rev = "e"+"h"= "eh", rev = "l" + "eh" = "olleh"
print("Reversed =", rev)
```

```python
s = input("Enter string: ") # check palindrome
rev = ""
for ch in s:
    rev = ch + rev
if s == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Print numbers from 1 to 10 except 5 (while loop)
i = 1

while i <= 10:
    if i != 5:
        print(i)
    i += 1
```

```python
# Print digits of a number in reverse using while loop
num = int(input("Enter a number: "))

while num > 0:
    digit = num % 10       # last digit
    print(digit, end=" ")
    num = num // 10        # remove last digit
```
